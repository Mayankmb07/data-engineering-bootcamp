```sql
SELECT * FROM sales_data;
SELECT * FROM customers;
```

```sql
SELECT
    c.name,
    s.product,
    s.region
FROM sales_data s INNER JOIN customers c ON s.customer_id = c.customer_id;
```

```sql
SELECT
    c.name,
    COUNT(s.product) AS No_Of_Product_Purchased
FROM customers c JOIN sales_data s on c.customer_id = s.customer_id
GROUP BY c.name
ORDER BY No_Of_Product_Purchased DESC ;
```

```sql
SELECT
    c.name,
    s.amount
FROM customers c JOIN sales_data s on c.customer_id = s.customer_id
WHERE c.gender = 'F' AND s.amount > 1000 ;
```

```sql
SELECT
    c.city,
    SUM(s.amount) AS Total_Spend
FROM customers c JOIN sales_data s on c.customer_id = s.customer_id
GROUP BY c.city
ORDER BY Total_Spend DESC
```
