```sql
SELECT
  region,
  count(*),
  sum(amount)
FROM sales_data
GROUP BY region
```

```sql
SELECT *
FROM sales_data
WHERE sale_date > '2023-10-02'
```

```sql
SELECT
customer_id,
amount
FROM sales_data
WHERE amount > 2000
ORDER BY amount DESC
```

```sql
SELECT
    region,
    MAX(amount) as single_sale_amount
FROM sales_data
GROUP BY region
```
