```sql
SELECT 
    l.application_id,
    l.customer_id,
    l.loan_amount_requested,
    c.customer_name,
    l.loan_status
FROM customer_details c 
JOIN loan_application L ON c.customer_id = L.customer_id
WHERE l.loan_status = 'Approved';
```

```sql
SELECT
    C.customer_name,
    L.application_id,
    L.loan_status
FROM customer_details C 
LEFT JOIN loan_application L ON C.customer_id = L.customer_id;
```

```sql
SELECT
    C.customer_id,
    C.customer_name,
    L.application_id,
    L.loan_status
FROM customer_details C 
LEFT JOIN loan_application L ON C.customer_id = L.customer_id
WHERE L.application_id IS NULL;
```

```sql
SELECT
    c.customer_name,
    c.city,
    l.loan_type,
    l.loan_amount_requested
FROM customer_details c 
JOIN loan_application l on c.customer_id = l.customer_id;
```

```sql
--2
SELECT
    c.customer_name,
    c.city,
    l.loan_type,
    l.loan_amount_requested
FROM customer_details c 
JOIN loan_application l on c.customer_id = l.customer_id
WHERE l.loan_type = 'Car Loan' AND c.city = 'Pune';
```

```sql
--3
SELECT
    c.city,
    ROUND(AVG(l.loan_amount_requested)) AS avg_loan_amount
FROM customer_details c 
JOIN loan_application l on c.customer_id = l.customer_id
GROUP BY c.city
ORDER BY avg_loan_amount DESC;
```

```sql
SELECT
    c.customer_name,
    COALESCE(COUNT(l.application_id),0) AS total_applications
FROM customer_details c 
LEFT JOIN loan_application l on c.customer_id = l.customer_id
GROUP BY c.customer_name ;
```
